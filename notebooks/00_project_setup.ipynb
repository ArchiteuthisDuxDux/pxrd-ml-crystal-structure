## Внешние raw-источники

В проекте хранятся только результаты обработки; raw-источники остаются в исходной папке.

```python
import pandas as pd

sources = pd.DataFrame([
    {'source': name, 'path': str(path), 'exists': Path(path).exists()}
    for name, path in RAW_SOURCES.items()
])
display(sources)
if not sources['exists'].all():
    raise FileNotFoundError('Недоступны raw-источники: ' + ', '.join(sources.loc[~sources.exists, 'source']))
```
